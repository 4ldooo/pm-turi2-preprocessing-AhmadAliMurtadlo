```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
df
```

```python
df.isna().sum()
```

```python
df['usia'] = df['usia'].fillna(df['usia'].median())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
df
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
X
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2})
X = pd.get_dummies(X, columns=['kota'], dtype=int)
y = y.map({'Tidak':0,'Ya':1})
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
print("X_train:", X_train.shape, "| X_test:", X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
num = ['usia','pendapatan','pendidikan']
sc = StandardScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num]  = sc.transform(X_test[num])  # cegah data leakage
X_train
```

```python
print("Nilai hilang di X_train:", X_train.isna().sum().sum())
print("Nilai hilang di X_test :", X_test.isna().sum().sum())
print("Semua kolom numerik?", all(np.issubdtype(dt, np.number) for dt in X_train.dtypes))
print("\nMean kolom numerik X_train (harus ~0):")
print(X_train[num].mean().round(4))
print("\nStd kolom numerik X_train (harus ~1):")
print(X_train[num].std(ddof=0).round(4))
```

## 4. Jelaskan mengapa `fit_transform` pada data latih tetapi hanya `transform` pada data uji

`fit_transform` melakukan dua hal sekaligus: **fit** (menghitung parameter statistik seperti mean & std untuk `StandardScaler`, atau min & max untuk `MinMaxScaler`) dan **transform** (menerapkan parameter tersebut ke data). Parameter ini **hanya boleh dihitung dari data latih**, karena:

- Data uji harus diperlakukan seolah-olah "belum pernah dilihat" oleh model, meniru kondisi dunia nyata saat model menerima data baru.
- Jika kita ikut menghitung mean/std/min/max dari data uji (misalnya dengan memanggil `fit_transform` lagi di data test), maka informasi dari data uji akan "bocor" ke proses training — ini disebut **data leakage**.
- Data leakage membuat evaluasi model menjadi bias (terlihat lebih baik dari performa sebenarnya), sehingga hasil di dunia nyata bisa mengecewakan.

Karena itu, urutan yang benar adalah: **split terlebih dahulu → fit scaler hanya di data train → transform data train dan data test memakai scaler yang sama.**

# E. Refleksi

**1. Mengapa urutan 'split dulu, baru scaling' penting?**

Karena jika scaling dilakukan sebelum split (fit pada seluruh dataset), statistik (mean, std, min, max) yang digunakan untuk scaling akan mengandung informasi dari data uji. Ini menyebabkan **data leakage**: model "mengintip" karakteristik data yang seharusnya belum ia lihat, sehingga evaluasi performa model menjadi tidak jujur/terlalu optimis. Dengan split dulu, scaler hanya belajar dari data latih, menjaga data uji benar-benar independen sebagai simulasi data baru.

**2. Kapan label encoding dan kapan one-hot encoding digunakan?**

- **Label encoding** (memetakan kategori ke angka urut, misal SMA=0, S1=1, S2=2) digunakan untuk data **ordinal**, yaitu kategori yang memiliki urutan/tingkatan yang bermakna. Contoh: tingkat pendidikan, karena S2 memang "lebih tinggi" dari S1 dan SMA.
- **One-hot encoding** (membuat kolom biner 0/1 untuk tiap kategori) digunakan untuk data **nominal**, yaitu kategori tanpa urutan yang bermakna. Contoh: kota (Bandung, Jakarta, Surabaya) — tidak ada kota yang "lebih besar" dari kota lain secara numerik, sehingga label encoding di sini justru akan menyesatkan model (model bisa salah menganggap ada hubungan ordinal padahal tidak ada).

**3. Apa perbedaan normalisasi dan standardisasi?**

- **Normalisasi** (contoh: `MinMaxScaler`) mengubah skala data ke rentang tertentu, biasanya [0, 1], dengan rumus `(x - min) / (max - min)`. Cocok digunakan saat distribusi data tidak diketahui normal, atau saat algoritma membutuhkan rentang nilai yang tetap (misalnya jaringan saraf/neural network, atau algoritma berbasis jarak seperti KNN).
- **Standardisasi** (contoh: `StandardScaler`) mengubah data agar memiliki rata-rata (mean) = 0 dan standar deviasi = 1, dengan rumus `(x - mean) / std`. Cocok digunakan saat data mendekati distribusi normal, atau saat algoritma mengasumsikan data terdistribusi normal (misalnya regresi linear, SVM, PCA).
- Perbedaan utama: normalisasi membatasi nilai ke rentang tertentu (sensitif terhadap outlier karena bergantung pada min/max), sedangkan standardisasi berpusat pada distribusi data (rata-rata & sebaran) dan tidak membatasi rentang nilai secara eksplisit.
